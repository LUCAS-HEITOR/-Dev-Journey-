## 1. The Python Object Model & Type System

### 💡 Concept
In Python, **everything is an object**. Variables in Python are not typed memory slots (like in C); they are simply **named references (pointers)** pointing to objects in the heap.

- **Dynamic Typing:** Variable types are resolved at runtime, not compile-time.
- **Strong Typing:** Python strictly prohibits implicit, unsafe operations across incompatible types (e.g., `'5' + 5` raises `TypeError`).

```python
import sys

# 1.1 Object Inspection
x = 42
print('Value      :', x)
print('Type       :', type(x))
print('Memory ID  :', hex(id(x)))
print('Memory Size:', sys.getsizeof(x), 'bytes')

# 1.2 Identity (is) vs Equality (==)
# '==' checks if values are equivalent
# 'is' checks if both variables point to the EXACT same memory address
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

print('\nlist_a == list_b (Value match)    :', list_a == list_b)  # True
print('list_a is list_b (Same memory ID) :', list_a is list_b)  # False
print('list_a is list_c (Same pointer)   :', list_a is list_c)  # True
```

### 1.3 Mutability vs Immutability: The Fundamental Axis

| Category | Types | Characteristics |
| :--- | :--- | :--- |
| **Immutable** | `int`, `float`, `complex`, `str`, `tuple`, `frozenset`, `bytes`, `bool` | Value cannot be modified in-place after creation. Modifications create a **new object** in RAM. |
| **Mutable** | `list`, `dict`, `set`, `bytearray`, custom class instances | Value can be altered directly in memory without changing object ID. |

```python
# Immutable Mutation Attempt: Generates a NEW memory address
text = 'Hello'
orig_id = id(text)
text += ' World'
new_id = id(text)
print('String ID changed after modification:', orig_id != new_id)  # True

# Mutable In-Place Mutation: Preserves the EXACT SAME memory address
numbers = [1, 2]
orig_list_id = id(numbers)
numbers.append(3)
new_list_id = id(numbers)
print('List ID unchanged after mutation   :', orig_list_id == new_list_id)  # True
```

## 2. Numeric Types

Python supports 4 primary numeric classes: `int`, `float`, `complex`, and the high-precision `decimal.Decimal` and `fractions.Fraction`.

```python
import math
from decimal import Decimal
from fractions import Fraction

# 2.1 int: Arbitrary-precision (no 32-bit/64-bit overflow limitations!)
huge_number = 2 ** 256
print(f'2^256 = {huge_number}')
print('Bit length of 2^256:', huge_number.bit_length(), 'bits')

# Number representations
print('Hex 0xFF   :', 0xFF)
print('Binary 0b1010:', 0b1010)

# 2.2 float: IEEE 754 Floating-Point Precision Gotcha
imprecision = 0.1 + 0.2
print(f'\n0.1 + 0.2 = {imprecision:.17f}')
print('0.1 + 0.2 == 0.3 is:', 0.1 + 0.2 == 0.3)  # False!
print('math.isclose(0.1 + 0.2, 0.3):', math.isclose(0.1 + 0.2, 0.3))  # True

# 2.3 High-Precision Financial Math with Decimal
dec_a = Decimal('0.1')
dec_b = Decimal('0.2')
print('Decimal Exact Sum:', dec_a + dec_b)  # Exactly 0.3

# 2.4 Exact Rational Fractions
frac1 = Fraction(1, 3)
frac2 = Fraction(1, 6)
print('Fraction 1/3 + 1/6:', frac1 + frac2)  # 1/2

# 2.5 complex numbers
z = 3 + 4j
print(f'Complex z: {z} | Real: {z.real} | Imag: {z.imag} | Conjugate: {z.conjugate()}')
```

## 3. Booleans & Truth Value Testing (Truthiness)

- In Python, `bool` is an explicit subclass of `int` (`issubclass(bool, int) == True`).
- `True == 1` and `False == 0` evaluate to `True`.

### Falsy Values in Python:
`None`, `False`, `0`, `0.0`, `0j`, `Decimal(0)`, `Fraction(0, 1)`, empty collections (`''`, `()`, `[]`, `{}`, `set()`, `range(0)`), and objects whose `__bool__()` or `__len__()` returns `False` or `0`.

```python
# Boolean is an integer subclass
print('True + True + False =', True + True + False)  # 2
print('isinstance(True, int):', isinstance(True, int))  # True

# Testing Truthiness
def inspect_truthiness(val):
    status = 'TRUTHY' if val else 'FALSY'
    print(f'{repr(val):<20} -> {status}')

inspect_truthiness([])
inspect_truthiness([0])
inspect_truthiness('')
inspect_truthiness(' ')
inspect_truthiness(0)
inspect_truthiness(None)

# Short-Circuit Evaluation
# 'or' returns the first truthy operand, or the last operand
# 'and' returns the first falsy operand, or the last operand
user_name = '' or 'Anonymous'
print('\nResolved username via short-circuit:', user_name)
```

## 4. Sequence Types

Sequences represent ordered collections indexed by non-negative integers. They support indexing `seq[i]`, slicing `seq[start:stop:step]`, membership testing `in`, concatenation `+`, repetition `*`, and `len()`.

```python
# 4.1 str (Immutable Unicode Sequence)
greeting = 'Python Data Types'
print('Slicing with step [::2] :', greeting[::2])
print('Reversing string [::-1]  :', greeting[::-1])
print('Upper & Replace          :', greeting.upper().replace('TYPES', 'STRUCTURES'))

# 4.2 bytes & bytearray
binary_data = b'Hello\x00\xFF'
mutable_buffer = bytearray(binary_data)
mutable_buffer[0] = ord('J')
print('Mutable bytearray:', mutable_buffer)

# 4.3 range: O(1) Memory sequence
huge_range = range(0, 1_000_000_000, 5)
print(f'Range object size: {sys.getsizeof(huge_range)} bytes (constant size for 1 billion items!)')
print('Index lookup in range:', huge_range[10])
```

```python
from collections import namedtuple
from typing import NamedTuple

# 4.4 list (Mutable Dynamic Array) vs tuple (Immutable Record)
matrix = [10, 20, 30, 40, 50]
matrix[1:4] = [99]  # Slice assignment
print('Modified list after slice assignment:', matrix)

# NamedTuple: Memory-efficient immutable records with named fields
class GeoPoint(NamedTuple):
    latitude: float
    longitude: float
    label: str = 'Waypoint'

point = GeoPoint(-23.5505, -46.6333, 'São Paulo')
print('\nNamedTuple:', point)
print('Access by attribute:', point.latitude, point.longitude)
print('Access by index    :', point[0])
```

## 5. Set Types: `set` & `frozenset`

Sets are **unordered collections of unique, hashable objects** backed by a hash table.
- $O(1)$ average time complexity for membership testing (`x in s`), additions, and removals.
- Mathematical operations: Union `|`, Intersection `&`, Difference `-`, Symmetric Difference `^`.

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print('Union (|)               :', set_a | set_b)
print('Intersection (&)        :', set_a & set_b)
print('Difference (A - B)      :', set_a - set_b)
print('Symmetric Difference (^):', set_a ^ set_b)

# frozenset: Immutable & Hashable Set
immutable_set = frozenset([1, 2, 3])

# Because frozenset is hashable, it CAN be used as a dictionary key or in another set!
lookup_table = {immutable_set: 'Cluster Alpha'}
print('\nFrozenset as dict key:', lookup_table[immutable_set])
```

## 6. Mapping Types: `dict` & Collections

Python's `dict` is a high-performance hash map that preserves **key insertion order** (standardized since Python 3.7).

```python
from collections import defaultdict, Counter

# 6.1 Dictionary Operations & Modern Merge Operators (PEP 584)
base_config = {'host': 'localhost', 'port': 8080, 'debug': True}
override_config = {'port': 9000, 'ssl': True}

merged = base_config | override_config  # Merge operator
print('Merged Dict (|):', merged)

# Dictionary Comprehension
squared_map = {x: x**2 for x in range(5)}
print('Dict Comprehension:', squared_map)

# 6.2 Specialized Collections
# defaultdict: Avoids KeyError by calling a factory on missing keys
grouped_items = defaultdict(list)
grouped_items['fruits'].append('Apple')
grouped_items['fruits'].append('Banana')
print('\ndefaultdict grouped:', dict(grouped_items))

# Counter: Frequency hash map
word_counts = Counter(['apple', 'banana', 'apple', 'orange', 'banana', 'apple'])
print('Top 2 Most Common:', word_counts.most_common(2))
```

## 7. Special Singleton Types: `NoneType`, `Ellipsis`, `NotImplemented`

- **`None` (`NoneType`):** Represents the absence of a value or null sentinel. Always compare with `is None`.
- **`...` (`Ellipsis`):** Used in type annotations (`Callable[..., int]`), NumPy multidimensional slicing, and code placeholders.
- **`NotImplemented`:** Returned by binary dunder methods (`__eq__`, `__add__`) to signal that an operation is unsupported with the given operand, allowing Python to try the reflected method.

```python
print('Type of None      :', type(None))
print('Type of Ellipsis  :', type(...))
print('Type of NotImpl   :', type(NotImplemented))

# Ellipsis in typing / slicing
def stub_function() -> ...:
    ...

print('Stub executed successfully with Ellipsis (...) body.')
```

## 8. Hashability & The Hash Table Contract

### 💡 What makes an object Hashable?
An object is **hashable** if:
1. It has a hash value that never changes during its lifetime (implements `__hash__()`).
2. It can be compared for equality with other objects (implements `__eq__()`).
3. If two objects compare equal (`a == b`), their hash values **MUST** be identical (`hash(a) == hash(b)`).

```python
# Immutable types produce hash integers
print('hash("Hello")      :', hash('Hello'))
print('hash((1, 2, 3))    :', hash((1, 2, 3)))
print('hash(42)           :', hash(42))

# Mutable objects are UNHASHABLE
try:
    hash([1, 2, 3])
except TypeError as err:
    print('\nCaught expected error with list:', err)

# ⚠️ Tuple containing a mutable item is also unhashable!
try:
    bad_tuple = (1, [2, 3])
    hash(bad_tuple)
except TypeError as err:
    print('Caught expected error with nested list inside tuple:', err)
```
