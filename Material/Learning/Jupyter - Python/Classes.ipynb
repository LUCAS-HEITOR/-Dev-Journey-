## 1. What are Classes and Objects?

- **Class:** The blueprint or template of an entity.
- **Object:** An instance created from a class.
- **Attribute:** A property or piece of data belonging to an object.
- **Method:** An action or function that an object knows how to perform.

For example, `Dog` can be a class. Each individual dog will be an object (instance) of that class.

```python
class Dog:
    pass


rex = Dog()
max_dog = Dog()

print(type(rex))
print(rex is max_dog)
print(rex.__class__.__name__)
```

## 2. Attributes and Methods

Attributes store state and data. Methods define behavior.

The `self` parameter represents the specific instance of the object and must appear as the first parameter in instance methods.

```python
class Dog:
    species = "Canis familiaris"

    def bark(self):
        return "Woof woof!"


rex = Dog()
rex.name = "Rex"
rex.age = 3

print(rex.name)
print(rex.age)
print(rex.species)
print(rex.bark())
```

## 3. The `__init__` Constructor

`__init__` is executed automatically when an object is created. It initializes the object's instance attributes.

The `self` keyword allows values to be bound to the current instance.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name} and I am {self.age} years old."


person_one = Person("Alice", 25)
person_two = Person("Bob", 30)

print(person_one.introduce())
print(person_two.introduce())
```

## 4. Special Methods (Dunder Methods)

Special methods have names surrounded by double underscores, such as `__str__` and `__len__`. They define how objects interact with Python's built-in functions, operators, and string formatting.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: ${self.price:.2f}"

    def __len__(self):
        return len(self.name)


product = Product("Keyboard", 129.90)
print(product)
print("Characters in name:", len(product))
```

## 5. Encapsulation with `@property`

In Python, we can manage attribute access and modification cleanly using `@property` getters and setters.

This is especially useful for validating values before updating internal state.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError("Insufficient funds.")
        self._balance -= amount


account = BankAccount("Alice", 100)
account.deposit(50)
account.withdraw(30)
print(f"Owner: {account.owner} | Balance: ${account.balance:.2f}")
```

## 6. Class Methods and Static Methods

- `@classmethod` receives `cls` as its first argument and can access/modify class state or serve as alternative constructors.
- `@staticmethod` does not receive `self` or `cls`; it behaves like a regular function grouped inside the class namespace for logical organization.

```python
class Temperature:
    default_scale = "Celsius"

    def __init__(self, celsius):
        self.celsius = celsius

    @classmethod
    def from_fahrenheit(cls, fahrenheit):
        celsius = (fahrenheit - 32) * 5 / 9
        return cls(celsius)

    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15


temp = Temperature.from_fahrenheit(68)
print(f"Celsius: {temp.celsius:.1f}")
print(f"Kelvin: {Temperature.celsius_to_kelvin(20):.2f}")
```

## 7. Inheritance

Inheritance allows a new class to reuse and extend attributes and methods from an existing parent class.

- **Parent class (Base/Superclass):** Provides shared, generic behavior.
- **Child class (Subclass):** Inherits and can add or override behaviors.
- `super()` grants access to methods in the parent class.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        return "Generic sound"


class Cat(Animal):
    def make_sound(self):
        return "Meow!"


class Dog(Animal):
    def make_sound(self):
        return "Woof woof!"


animals = [Cat("Mimi"), Dog("Rex")]
for animal in animals:
    print(f"{animal.name} says: {animal.make_sound()}")
```

## 8. Polymorphism

Polymorphism allows objects of different classes to respond to the same method interface in their own distinct ways.

In the previous example, all animal subclasses have a `make_sound()` method, but each class exhibits its own specialized sound.

## 9. Exercises

Try solving these challenges before reviewing the solution below:

1. Create a `Book` class with `title` and `author` attributes.
2. Add a `description()` method that returns a formatted summary of the book.
3. Create an `EBook` class that inherits from `Book` and introduces a `file_size_mb` attribute.
4. Create a `Rectangle` class with methods to calculate `area` and `perimeter`.
5. Validate that width and height are strictly positive numbers.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def description(self):
        return f"'{self.title}' by {self.author}"


class EBook(Book):
    def __init__(self, title, author, file_size_mb):
        super().__init__(title, author)
        self.file_size_mb = file_size_mb

    def description(self):
        return f"{super().description()} | file size: {self.file_size_mb} MB"


class Rectangle:
    def __init__(self, width, height):
        if width <= 0 or height <= 0:
            raise ValueError("Width and height must be positive.")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


books = [
    Book("Clean Code", "Robert C. Martin"),
    EBook("Fluent Python", "Luciano Ramalho", 4.5),
]
for book in books:
    print(book.description())

rect = Rectangle(5, 3)
print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

## 10. Summary & Key Takeaways

- Use `class` to define a template for creating objects.
- Use `__init__` to initialize instance state.
- Use `self` to access the current object instance.
- Use methods to represent behavior.
- Use `@property` for clean data encapsulation and validation.
- Use inheritance to share and specialize behaviors across related classes.
- Use polymorphism when different objects should respond to the same interface.

### When should you use classes?

Use classes when data and functions naturally belong to the same entity, when creating multiple instances with shared structure, or when modeling real-world domain logic.
