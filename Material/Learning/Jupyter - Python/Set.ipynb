## 1. What is a `set`?

A `set` represents an unordered collection of distinct elements:

- Does not allow duplicate items;
- Does not guarantee insertion or display order;
- Accepts only **hashable** elements (immutable types such as numbers, strings, and frozen tuples);
- Highly optimized for $O(1)$ average-time membership lookups (`in` operator).

It directly mirrors the mathematical concept of a set.

```python
# A set can be created using curly braces
fruits = {"banana", "apple", "orange", "banana"}
print(fruits)
print(len(fruits))  # 'banana' appears only once

# To create an empty set, use set(), not {}
empty_set = set()
empty_dict = {}
print(type(empty_set))
print(type(empty_dict))
```

## 2. Creating Sets from Other Iterables

Use `set(iterable)` to convert a list, tuple, or string into a set. This is a common technique to eliminate duplicate entries.

Note: `set("banana")` splits the string into distinct characters, not words.

```python
numbers = set([1, 2, 2, 3, 4, 4])
letters = set("banana")
courses = {"Python", "SQL", "Python", "Git"}

print(numbers)
print(letters)
print(courses)

# Deduplicating while preserving original list order:
def remove_duplicates_ordered(seq):
    return list(dict.fromkeys(seq))

names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
print(remove_duplicates_ordered(names))
```

## 3. Adding, Removing, and Clearing Elements

- `add(item)`: Adds a single element.
- `update(iterable)`: Adds multiple elements from an iterable.
- `remove(item)`: Removes an item, raising `KeyError` if it is not present.
- `discard(item)`: Removes an item without raising an error if absent.
- `pop()`: Removes and returns an arbitrary element.
- `clear()`: Empties the entire set.

```python
languages = {"Python", "Java"}
languages.add("C++")
languages.update(["Go", "Rust", "Python"])
print(languages)

languages.remove("Java")
languages.discard("Haskell")  # No error raised if element does not exist
print(languages)

removed_item = languages.pop()
print("Removed item:", removed_item)
print("Current set:", languages)

clone = languages.copy()
clone.clear()
print("Cleared clone:", clone)
print("Original preserved:", languages)
```

## 4. Querying and Membership Testing

The `item in set` lookup is one of the biggest advantages of `set`, operating in $O(1)$ average time.

Sets do not support integer indexing (e.g., `s[0]` will raise `TypeError`). To iterate through elements, use a `for` loop.

```python
permissions = {"read", "write", "execute"}

print("read" in permissions)
print("delete" not in permissions)

for perm in sorted(permissions):
    print(perm)

# len returns the number of unique elements
print("Count:", len(permissions))
```

## 5. Mathematical Set Operations

Given two sets:

- **Union** (`|` or `union()`): All unique elements from both sets.
- **Intersection** (`&` or `intersection()`): Only elements present in both sets.
- **Difference** (`-` or `difference()`): Elements in the first set but not in the second.
- **Symmetric Difference** (`^` or `symmetric_difference()`): Elements in either set, but not in both.

```python
frontend = {"HTML", "CSS", "JavaScript"}
backend = {"Python", "SQL", "JavaScript"}

print("Union:", frontend | backend)
print("Intersection:", frontend & backend)
print("Frontend only:", frontend - backend)
print("Exclusive to one:", frontend ^ backend)

# Method versions are equivalent
print(frontend.union(backend))
print(frontend.intersection(backend))
print(frontend.difference(backend))
print(frontend.symmetric_difference(backend))
```

## 6. Comparing Sets

Use these methods and operators to check relationships between sets:

- `issubset()` or `<=`: Every element of A is also in B;
- `issuperset()` or `>=`: Set A contains every element of B;
- `isdisjoint()`: Sets have no common elements;
- `==`: Sets contain identical elements regardless of order.

```python
core_langs = {"Python", "JavaScript"}
web_langs = {"HTML", "CSS", "JavaScript", "Python"}
db_langs = {"SQL", "NoSQL"}

print(core_langs.issubset(web_langs))
print(web_langs.issuperset(core_langs))
print(web_langs.isdisjoint(db_langs))
print({1, 2, 3} == {3, 2, 1})
```

## 7. In-Place Updates

The following methods mutate the existing set in-place:

- `intersection_update()`: Retains only elements found in both sets;
- `difference_update()`: Removes all elements found in the specified set;
- `symmetric_difference_update()`: Keeps only elements unique to each set.

If you need to preserve the original set, make a `.copy()` or use the binary operators (`&`, `-`, `^`).

```python
attendees = {"Alice", "Bob", "Charlie"}
checked_in = {"Alice", "Charlie", "David"}

all_registered = attendees | checked_in
print("All registered:", all_registered)

attendees.intersection_update(checked_in)
print("Present attendees:", attendees)

missing = {"Alice", "Bob", "Charlie"}
missing.difference_update(checked_in)
print("Missing:", missing)
```

## 8. Mutable `set` vs Immutable `frozenset`

Standard `set` is mutable and therefore unhashable (cannot be used as a dictionary key or placed inside another set).

`frozenset` is the immutable counterpart: it is hashable and can be used as a dictionary key or stored within sets.

```python
workout_days = frozenset(["Mon", "Wed", "Fri"])
schedule = {workout_days: "Strength Training"}

print(workout_days)
print(schedule[workout_days])

# Lists are mutable and cannot be elements of a set:
try:
    set([[1, 2], [3, 4]])
except TypeError as err:
    print("Expected error:", err)
```

## 10. Summary: Choosing the Right Data Structure

| Data Structure | Allows Duplicates? | Ordered / Indexed? | Mutable? | Primary Use Case |
|---|---:|---:|---:|---|
| `list` | Yes | Yes (by integer index) | Yes | Sequential collections |
| `tuple` | Yes | Yes (by integer index) | No | Immutable records / coordinates |
| `set` | No | No | Yes | Fast membership & mathematical sets |
| `frozenset` | No | No | No | Hashable immutable set (dict keys) |
| `dict` | No (keys must be unique) | Yes (insertion order) | Yes | Key-value mappings |

### Quick Checklist

- Need to eliminate duplicates? Use `set`.
- Need to deduplicate while preserving order? Use `dict.fromkeys()`.
- Need positional indexing? Use `list` or `tuple`.
- Need a set as a dictionary key? Use `frozenset`.
- Need group comparisons (overlaps, differences)? Use `|`, `&`, `-`, and `^`.
