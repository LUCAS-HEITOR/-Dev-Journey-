## 1. Print & Comments

- `print()` displays output to the console.
- `#` starts a single-line comment.
- `"""..."""`  or `'''...'''` for multi-line comments / docstrings.

```python
# This is a comment — Python ignores it
print("Hello, World!")
print("Welcome to Python!")

# Print multiple values separated by a space
print("Name:", "Lucas", "| Age:", 22)
```

## 2. Variables & Data Types

Python is **dynamically typed** — no need to declare the type.

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42` | Whole number |
| `float` | `3.14` | Decimal number |
| `str` | `"hello"` | Text |
| `bool` | `True` / `False` | Boolean |
| `NoneType` | `None` | Absence of value |

```python
name = "Lucas"         # str
age = 22               # int
height = 1.75          # float
is_student = True      # bool
nickname = None        # NoneType

print(type(name), name)
print(type(age), age)
print(type(height), height)
print(type(is_student), is_student)
print(type(nickname), nickname)
```

```python
# Multiple assignment
x, y, z = 10, 20, 30
print(x, y, z)

# Swap variables (no temp needed!)
a, b = 1, 2
a, b = b, a
print(f"After swap: a={a}, b={b}")
```

## 3. Strings

Strings are sequences of characters. They are **immutable** (cannot be changed in place).

```python
greeting = "Hello, World!"

# Indexing (0-based)
print("First char:", greeting[0])
print("Last char:", greeting[-1])

# Slicing [start:stop:step]
print("First 5:", greeting[:5])
print("Reversed:", greeting[::-1])

# Length
print("Length:", len(greeting))
```

```python
text = "  Python is Awesome  "

print(text.strip())          # Remove whitespace
print(text.strip().lower())  # Lowercase
print(text.strip().upper())  # Uppercase
print(text.strip().replace("Awesome", "Fun"))
print(text.strip().split())  # Split into a list of words
print("Python" in text)      # Check if substring exists
```

## 4. Numbers & Math

| Operator | Meaning | Example |
|:---:|---|---|
| `+` | Addition | `5 + 3 → 8` |
| `-` | Subtraction | `5 - 3 → 2` |
| `*` | Multiplication | `5 * 3 → 15` |
| `/` | Division (float) | `7 / 2 → 3.5` |
| `//` | Floor division | `7 // 2 → 3` |
| `%` | Modulus | `7 % 2 → 1` |
| `**` | Exponent | `2 ** 3 → 8` |

```python
print("7 / 2  =", 7 / 2)    # Float division
print("7 // 2 =", 7 // 2)   # Floor division
print("7 % 2  =", 7 % 2)    # Remainder
print("2 ** 10 =", 2 ** 10)  # Exponentiation

# Built-in math functions
print("abs(-5):", abs(-5))
print("round(3.7):", round(3.7))
print("max(1, 5, 3):", max(1, 5, 3))
print("min(1, 5, 3):", min(1, 5, 3))
```

## 5. Type Conversion

Convert between types using `int()`, `float()`, `str()`, `bool()`.

```python
# String to number
num_str = "42"
num_int = int(num_str)
print(num_int + 8)  # 50

# Number to string
age = 22
print("I am " + str(age) + " years old.")

# Truthiness
print(bool(0))     # False
print(bool(1))     # True
print(bool(""))    # False (empty string)
print(bool("hi"))  # True (non-empty string)
```

## 7. Conditionals (if / elif / else)

Python uses **indentation** (not braces) to define blocks.

**Comparison:** `==`, `!=`, `<`, `>`, `<=`, `>=`

**Logical:** `and`, `or`, `not`

```python
score = 85

if score >= 90:
    print("Grade: A (Excellent!)")
elif score >= 70:
    print("Grade: B (Great!)")
elif score >= 50:
    print("Grade: C (Okay)")
else:
    print("Grade: F (Try again)")
```

```python
# Logical operators
temp = 25
is_sunny = True

if temp > 20 and is_sunny:
    print("Perfect day!")

# Ternary expression (one-liner)
status = "adult" if age >= 18 else "minor"
print(f"Status: {status}")
```

## 8. Loops (for, while)

### `for` loop — iterate over sequences

```python
# Iterate over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")
```

```python
# range(start, stop, step)
for i in range(1, 6):
    print(f"Count: {i}")

# Even numbers from 0 to 10
print("Even:", list(range(0, 11, 2)))
```

### `while` loop — repeat while condition is true

```python
countdown = 5
while countdown > 0:
    print(countdown, end="... ")
    countdown -= 1
print("Liftoff! 🚀")
```

```python
# break and continue
for i in range(10):
    if i == 3:
        continue  # Skip 3
    if i == 7:
        break     # Stop at 7
    print(i, end=" ")
print()  # newline
```

```python
# enumerate — get index AND value
languages = ["C++", "Python", "Lua"]
for idx, lang in enumerate(languages, start=1):
    print(f"{idx}. {lang}")
```

## 9. Lists

Lists are **ordered, mutable** collections. The most used data structure in Python.

```python
numbers = [10, 20, 30, 40, 50]

# Accessing elements
print("First:", numbers[0])
print("Last:", numbers[-1])
print("Slice [1:4]:", numbers[1:4])

# Modifying
numbers.append(60)       # Add to end
numbers.insert(0, 5)     # Insert at index 0
numbers.remove(30)       # Remove by value
popped = numbers.pop()   # Remove & return last

print("List:", numbers)
print("Popped:", popped)
print("Length:", len(numbers))
```

```python
# Sorting
scores = [85, 92, 78, 95, 88]
scores.sort()                # In-place sort (ascending)
print("Sorted:", scores)

scores.sort(reverse=True)    # Descending
print("Descending:", scores)

# Useful functions
print("Sum:", sum(scores))
print("Min:", min(scores))
print("Max:", max(scores))
```

## 10. Tuples & Unpacking

Tuples are like lists but **immutable** (cannot be modified after creation).

```python
# Tuple
coordinates = (10, 20)
print("X:", coordinates[0])
print("Y:", coordinates[1])

# Unpacking
x, y = coordinates
print(f"Unpacked: x={x}, y={y}")

# Tuple as function return
def min_max(numbers):
    return min(numbers), max(numbers)

lo, hi = min_max([3, 1, 4, 1, 5, 9])
print(f"Min: {lo}, Max: {hi}")
```

## 11. Dictionaries

Key-value pairs with **O(1) average lookups**. Keys must be hashable (immutable).

```python
student = {
    "name": "Lucas",
    "age": 22,
    "languages": ["C++", "Python", "Lua"]
}

print("Name:", student["name"])
print("Languages:", student["languages"])

# Safe access with .get()
print("GPA:", student.get("gpa", "Not set"))

# Add / modify
student["gpa"] = 3.8
print("Updated:", student)
```

```python
# Iterating a dictionary
for key, value in student.items():
    print(f"  {key}: {value}")

print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
```

## 12. Functions

Functions are defined with `def`. They can have default parameters, return values, and type hints.

```python
def greet(name, greeting="Hello"):
    """Greets a person with a custom or default greeting."""
    return f"{greeting}, {name}!"

print(greet("Lucas"))
print(greet("Alice", "Hey"))
```

```python
# *args and **kwargs
def summarize(*args, **kwargs):
    print("Positional args:", args)
    print("Keyword args:", kwargs)

summarize(1, 2, 3, lang="Python", level="beginner")
```

```python
# Lambda (anonymous function)
square = lambda x: x ** 2
print("5² =", square(5))

# Using lambda with sorted
students = [("Alice", 90), ("Bob", 75), ("Charlie", 95)]
ranked = sorted(students, key=lambda s: s[1], reverse=True)
print("Ranked:", ranked)
```

## 13. f-Strings & Formatting

f-Strings (Python 3.6+) are the cleanest way to format output.

```python
name = "Lucas"
score = 95.678

print(f"Name: {name}")
print(f"Score: {score:.2f}")          # 2 decimal places
print(f"Score: {score:>10.2f}")        # Right-aligned, 10 chars wide
print(f"Binary of 42: {42:b}")         # Binary representation
print(f"Hex of 255: {255:#x}")         # Hexadecimal
print(f"{'Centered':^20}")             # Center-aligned, 20 chars
```

## 14. List Comprehensions

A concise way to create lists from existing sequences.

Syntax: `[expression for item in iterable if condition]`

```python
# Squares of 1 to 10
squares = [x ** 2 for x in range(1, 11)]
print("Squares:", squares)

# Only even numbers
evens = [x for x in range(20) if x % 2 == 0]
print("Evens:", evens)

# Transform strings
words = ["hello", "world", "python"]
upper_words = [w.upper() for w in words]
print("Upper:", upper_words)
```

```python
# Dictionary comprehension
names = ["Alice", "Bob", "Charlie"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)

# Set comprehension
unique_lengths = {len(name) for name in names}
print("Unique lengths:", unique_lengths)
```

## 15. Error Handling (try / except)

Gracefully handle errors instead of crashing.

```python
# Basic try/except
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")

# Multiple exceptions
try:
    num = int("abc")
except ValueError as e:
    print(f"ValueError caught: {e}")

# try/except/else/finally
try:
    value = int("42")
except ValueError:
    print("Not a number")
else:
    print(f"Successfully converted: {value}")
finally:
    print("This always runs.")
```

## 16. File I/O Basics

Read and write files using `open()` with context managers (`with` statement).

```python
# Writing to a file
with open("example.txt", "w") as f:
    f.write("Line 1: Hello from Python!\n")
    f.write("Line 2: File I/O is simple.\n")

# Reading from a file
with open("example.txt", "r") as f:
    content = f.read()
    print(content)

# Cleanup
import os
os.remove("example.txt")
print("File created, read, and cleaned up.")
```
