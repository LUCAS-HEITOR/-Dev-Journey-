## 1. Hello World & Output

In C++, we use `#include <iostream>` to access input/output.

- `std::cout` prints text to the screen.
- `<<` is the **insertion operator** (sends data into the stream).
- `std::endl` or `\n` creates a new line.

````c++
#include <iostream>
````

````c++
std::cout << "Hello, World!" << std::endl;
````

````c++
// You can chain multiple values with <<
std::cout << "Line 1" << std::endl;
std::cout << "Line 2" << std::endl;
std::cout << "Result: " << 10 + 5 << std::endl;
````

## 2. Variables & Data Types

C++ is a **statically typed** language — every variable must have a declared type.

| Type | What it stores | Example |
|------|---------------|--------|
| `int` | Whole numbers | `42` |
| `double` | Decimal numbers | `3.14` |
| `char` | Single character | `'A'` |
| `bool` | True or False | `true` |
| `std::string` | Text | `"Hello"` |

````c++
int age = 22;
double height = 1.75;
char grade = 'A';
bool is_student = true;

std::cout << "Age: " << age << std::endl;
std::cout << "Height: " << height << " m" << std::endl;
std::cout << "Grade: " << grade << std::endl;
std::cout << "Student? " << is_student << std::endl; // 1 = true, 0 = false
````

````c++
// Strings require #include <string>
#include <string>

std::string name = "Lucas";
std::cout << "Name: " << name << std::endl;
std::cout << "Length: " << name.length() << " characters" << std::endl;
````

### `auto` keyword

Modern C++ can figure out the type automatically:

````c++
auto x = 10;       // int
auto pi = 3.14;    // double
auto msg = "Hi!";  // const char*

std::cout << x << ", " << pi << ", " << msg << std::endl;
````

### Constants

Use `const` for values that should **never change**:

````c++
const double PI = 3.14159265;
const int MAX_LIVES = 3;

std::cout << "PI = " << PI << std::endl;
std::cout << "Max Lives = " << MAX_LIVES << std::endl;

// PI = 5.0;  // ERROR! Cannot modify a const variable
````

## 4. Arithmetic Operators

| Operator | Meaning | Example |
|:---:|---|---|
| `+` | Addition | `5 + 3 = 8` |
| `-` | Subtraction | `5 - 3 = 2` |
| `*` | Multiplication | `5 * 3 = 15` |
| `/` | Division | `7 / 2 = 3` (integer!) |
| `%` | Modulus (remainder) | `7 % 2 = 1` |

````c++
int a = 10;
int b = 3;

std::cout << "a + b = " << a + b << std::endl;
std::cout << "a - b = " << a - b << std::endl;
std::cout << "a * b = " << a * b << std::endl;
std::cout << "a / b = " << a / b << " (integer division!)" << std::endl;
std::cout << "a % b = " << a % b << " (remainder)" << std::endl;
````

````c++
// To get decimal division, at least one operand must be double
double result = 10.0 / 3;
std::cout << "10.0 / 3 = " << result << std::endl;
````

## 5. Conditionals (if / else)

Control the flow of your program based on conditions.

**Comparison Operators:** `==`, `!=`, `<`, `>`, `<=`, `>=`

**Logical Operators:** `&&` (AND), `||` (OR), `!` (NOT)

````c++
int score = 85;

if (score >= 90) {
    std::cout << "Grade: A (Excellent!)" << std::endl;
} else if (score >= 70) {
    std::cout << "Grade: B (Great!)" << std::endl;
} else {
    std::cout << "Grade: C (Keep studying!)" << std::endl;
}
````

````c++
// Logical operators
int temp = 25;
bool is_sunny = true;

if (temp > 20 && is_sunny) {
    std::cout << "Perfect day to go outside!" << std::endl;
}

if (temp < 0 || temp > 40) {
    std::cout << "Extreme weather!" << std::endl;
} else {
    std::cout << "Temperature is normal." << std::endl;
}
````

## 6. Loops

### `for` loop — when you know how many times to repeat

````c++
// Classic for loop: initialization; condition; update
for (int i = 1; i <= 5; i++) {
    std::cout << "Count: " << i << std::endl;
}
````

### `while` loop — repeat while a condition is true

````c++
int countdown = 5;

while (countdown > 0) {
    std::cout << countdown << "... ";
    countdown--;
}
std::cout << "Liftoff! 🚀" << std::endl;
````

### Range-based `for` loop (Modern C++)

The cleanest way to iterate over a collection:

````c++
#include <vector>

std::vector<std::string> languages = {"C++", "Python", "Lua"};

for (const auto& lang : languages) {
    std::cout << "I'm learning: " << lang << std::endl;
}
````

## 7. Functions

Functions let you organize and reuse code.

```cpp
ReturnType function_name(ParameterType param) {
    // body
    return value;
}
```

Use `void` when the function doesn't return anything.

````c++
int add(int a, int b) {
    return a + b;
}

void greet(const std::string& name) {
    std::cout << "Hello, " << name << "! Welcome to C++." << std::endl;
}
````

````c++
std::cout << "3 + 7 = " << add(3, 7) << std::endl;
greet("Lucas");
````

````c++
// Function with default parameter
double calculate_area(double width, double height = 1.0) {
    return width * height;
}

std::cout << "Area (5 x 3): " << calculate_area(5, 3) << std::endl;
std::cout << "Area (5, default): " << calculate_area(5) << std::endl;
````

## 8. Arrays & Vectors

- **C-style array** (`int arr[5]`): Fixed size, determined at compile time.
- **`std::vector`**: Dynamic array that can grow and shrink. **Always prefer this!**

````c++
// C-style array (fixed size)
int scores[3] = {90, 85, 95};

std::cout << "First score: " << scores[0] << std::endl;
std::cout << "Third score: " << scores[2] << std::endl;
````

````c++
// std::vector — dynamic, safe, and modern
std::vector<int> nums = {10, 20, 30};

nums.push_back(40);  // Add element to the end
nums.push_back(50);

std::cout << "Size: " << nums.size() << std::endl;

for (const auto& n : nums) {
    std::cout << n << " ";
}
std::cout << std::endl;
````

````c++
// Useful vector operations
std::cout << "First element: " << nums.front() << std::endl;
std::cout << "Last element: "  << nums.back()  << std::endl;
std::cout << "Is empty? "      << nums.empty() << std::endl;

nums.pop_back();  // Remove last element
std::cout << "After pop_back, size: " << nums.size() << std::endl;
````

## 9. Strings

`std::string` is the standard way to work with text in C++.

````c++
std::string first = "Hello";
std::string second = "World";

// Concatenation
std::string greeting = first + ", " + second + "!";
std::cout << greeting << std::endl;

// Length
std::cout << "Length: " << greeting.length() << std::endl;

// Access individual characters
std::cout << "First char: " << greeting[0] << std::endl;
std::cout << "Last char: "  << greeting.back() << std::endl;
````

````c++
// Substring & find
std::string text = "Learning C++ is fun";

std::cout << "Substring (0, 8): " << text.substr(0, 8) << std::endl;

size_t pos = text.find("C++");
if (pos != std::string::npos) {
    std::cout << "Found 'C++' at position: " << pos << std::endl;
}
````
